```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np
import torch
```

```python
# Exemple de données (si tu n'as pas de DataFrame)
# Série temporelle simple
data = np.sin(np.linspace(0, 20, 200))

# Si data vient d'un DataFrame :
# data = df["value"].values

# -----------------------
# 1. Normalisation
# -----------------------
scaler = MinMaxScaler(feature_range=(0, 1))
data = np.array(data).reshape(-1, 1)      # sécurité : toujours (N,1)
data_scaled = scaler.fit_transform(data)

# -----------------------
# 2. Création des séquences
# -----------------------
def create_sequences(data, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length])
    return np.array(X), np.array(y)

seq_length = 15
X, y = create_sequences(data_scaled, seq_length)

# -----------------------
# 3. Conversion en tenseurs PyTorch
# -----------------------
X = torch.tensor(X, dtype=torch.float32)   # (batch, seq_len, 1)
y = torch.tensor(y, dtype=torch.float32)   # (batch, 1)

print("X shape:", X.shape)  # doit être (N, seq_length, 1)
print("y shape:", y.shape)  # doit être (N, 1)
```
