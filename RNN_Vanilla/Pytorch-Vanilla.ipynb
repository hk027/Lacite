```python
import torch.nn as nn
```

```python
class RNN(nn.Module) : 
    def __init__(self): 
        super(RNN,self).__init__()
        self.rnn = nn.Sequential(
            nn.RNN(input_size=28, hidden_size=128, batch_first=True),
            nn.RNN(input_size=128, hidden_size=128),
            nn.Linear(128, 10),
            nn.Softmax(dim=1)
        )
    
    def forward(self, x):
    #out = self.rnn(...)
    #Ici on envoie cette séquence dans :
    #self.rnn = nn.Sequential(
    #nn.RNN(28, 128, batch_first=True),
    #nn.RNN(128, 128),
    #nn.Linear(128, 10),
    #nn.Softmax(dim=1))
        out = self.rnn(x.view(-1, 28, 28)) # pernet de remettre l'image en forme
        return out
    
    #L’image devient une séquence
    #Elle passe par les deux RNN
    #Puis par la couche Linear
    #Puis par le Softmax
    #On obtient les probabilités pour les 10 classes (0 à 9)

model = RNN()
print(model)
```

<center>
Interprétation :

Chaque ligne de l’image (28 pixels) est vue comme un pas de temps.

Le premier RNN extrait une représentation temporelle de dimension 128.

Le second RNN approfondit cette représentation.

La couche Linear transforme l’état caché final en 10 sorties (une par classe).

Softmax convertit ces sorties en probabilités de classification.

1. batch_size

Par exemple :

   1. batch_size = 32 → le réseau traite 32 images en parallèle

   2. batch_size = 64 → 64 images en parallèle
