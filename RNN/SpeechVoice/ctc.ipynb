```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
# Suppose we have 2 classes: 'A', 'B', plus blank
num_classes = 3  # ['A', 'B', blank]
T = 5  # timesteps
N = 2  # batch size

# Sortie aléatoire du réseau (logits)
logits = torch.randn(T, N, num_classes).log_softmax(2)  # (T, N, C)

# Séquence cible : batch de taille 2
targets = torch.tensor([[0, 1], [1, 0]], dtype=torch.long)  # lettres indices

# Longueur de séquence pour chaque échantillon
input_lengths = torch.full((N,), T, dtype=torch.long)
target_lengths = torch.full((N,), 2, dtype=torch.long)

# Définir la perte CTC
ctc_loss = nn.CTCLoss(blank=2)  # blank index = 2

loss = ctc_loss(logits, targets, input_lengths, target_lengths)
print("CTC Loss:", loss.item())
```
