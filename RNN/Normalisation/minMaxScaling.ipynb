```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np
import pandas as pd
```

```python
# Données simulées (ex : températures, ventes, etc.)
data = pd.Series([12, 15, 18, 20, 22, 25, 27, 26, 24, 23, 21, 19])

scaler = MinMaxScaler(feature_range=(0, 1))
data_scaled = scaler.fit_transform(data.values.reshape(-1, 1))

print("Données originales :")
print(data.values)
print("Voici les donnees normalisees")
print(data_scaled)
```
