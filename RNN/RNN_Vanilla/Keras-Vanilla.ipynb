**Vanilla RNN** 

Dans sa forme la plus simple, appelée RNN classique, le réseau ne comporte qu'un seul état caché. Une formule de récurrence nous indique comment mettre à jour cet état caché en fonction de son état caché précédent et de l'entrée actuelle.

```python
from keras import layers, models
```

```python
def rnn(image_batch) : 
    # This defines a function rnn that takes a batch of images as input.Image_batch is expected to have shape (batch_size, 784) (flattened 28×28 images)
    # there are 28 time steps in total, with each time step receiving a 28-dimensional input vector. 
    x_image = layers.Reshape((28,28)) (image_batch)
    
    #Original input: (batch_size, 784) → each image is flattened.
    #Reshape((28,28)) converts it to (batch_size, 28, 28):
        ##28 time steps (sequence length)
        ##Each time step has 28 features (input size)
    #Idea: Treat each row of the image as a timestep for the RNN.
    # Here, return_sequences must be set to True for the subsequent RNN layers to function correctly.
    
    h_rnn1 = layers.SimpleRNN(units=128, return_sequences=True)(x_image)
    h_rnn2 = layers.SimpleRNN(units=128) (h_rnn1)
    _y = layers.Dense(10, activation='softmax') (h_rnn2)
    return _y

x  = layers.Input(shape=(784,))
y_ = layers.Input(shape=(10,))
y = rnn(x)

model = models.Model(x,y)
print(model.summary())
```
