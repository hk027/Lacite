```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from sklearn.preprocessing import MinMaxScaler
```

```python
# Exemple : données brutes (série temporelle)
# si tu as un DataFrame : data = df["value"].values
data = np.sin(np.linspace(0, 20, 200))   # exemple simple

# 1. Mise en forme correcte (N,1)
data = np.array(data).reshape(-1, 1)

# 2. Normalisation Min-Max → data_scaled
scaler = MinMaxScaler(feature_range=(0, 1))
data_scaled = scaler.fit_transform(data)
```

```python
def create_sequences(data, seq_length=30):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length])
    return np.array(X), np.array(y)

X, y = create_sequences(data_scaled)
X = torch.from_numpy(X).float()
y = torch.from_numpy(y).float()
X = X.unsqueeze(-1)  # [batch, seq_length, features]
```
