<h1>📊 Explication globale du code</h1>

<p>
Ce programme réalise un pipeline complet de <strong>AutoML</strong> avec 
<strong>H2O AutoML</strong>.
</p>

<h2>🔁 Étapes exécutées</h2>

<h3>1. Initialisation</h3>
<ul>
  <li>Démarre H2O (moteur ML distribué)</li>
</ul>

<h3>2. Chargement des données</h3>
<ul>
  <li>Dataset Titanic (classification : survivant ou non)</li>
</ul>

<h3>3. Préparation des variables</h3>
<ul>
  <li><code>y</code> = variable cible (survived)</li>
  <li><code>x</code> = variables explicatives</li>
</ul>

<h3>4. Split des données</h3>
<ul>
  <li>Entraînement (80%)</li>
  <li>Test (20%)</li>
</ul>

<h3>5. AutoML</h3>
<ul>
  <li>Teste plusieurs modèles automatiquement</li>
  <li>Optimise leurs hyperparamètres</li>
  <li>Classe les modèles (leaderboard)</li>
</ul>

<h3>6. Sélection du meilleur modèle</h3>
<ul>
  <li><code>leader</code> = modèle le plus performant</li>
</ul>

<h3>7. Évaluation</h3>
<ul>
  <li>Calcul des performances sur le test set</li>
</ul>

<hr>

<h2>🎯 Ce que fait AutoML</h2>
<ul>
  <li>Teste plusieurs algorithmes (Random Forest, GBM, XGBoost, Deep Learning)</li>
  <li>Optimise automatiquement les hyperparamètres</li>
  <li>Utilise la validation croisée</li>
  <li>Classe les modèles par performance</li>
</ul>

<p><strong>Résultat :</strong> le meilleur modèle est sélectionné automatiquement.</p>

```python
# Import de la bibliothèque principale H2O
import h2o

# Import du module AutoML de H2O pour automatiser le machine learning
from h2o.automl import H2OAutoML

# Démarrage du serveur H2O (local)
# Cela initialise l'environnement de calcul distribué
h2o.init()

# Chargement du dataset Titanic depuis une URL
# Les données sont automatiquement converties en H2OFrame (format H2O)
data = h2o.import_file("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv")

# Définition de la variable cible (ce qu'on veut prédire)
y = "survived"

# Récupération de toutes les colonnes du dataset
x = data.columns

# Suppression de la variable cible de la liste des variables explicatives
# (car on ne doit pas utiliser la réponse comme entrée)
x.remove(y)

# Division du dataset en deux parties :
# - 80% pour l'entraînement
# - 20% pour le test
train, test = data.split_frame(ratios=[0.8])

# Création d'un objet AutoML
# max_models=10 → limite le nombre de modèles entraînés
# seed=1 → permet de reproduire les résultats
aml = H2OAutoML(max_models=10, seed=1)

# Lancement de l'entraînement AutoML
# H2O va automatiquement :
# - tester plusieurs modèles (GBM, Random Forest, Deep Learning, etc.)
# - optimiser leurs hyperparamètres
# - sélectionner les meilleurs modèles
aml.train(x=x, y=y, training_frame=train)

# Récupération du meilleur modèle trouvé (leader du leaderboard)
leader = aml.leader

# Évaluation des performances du modèle sur les données de test
perf = leader.model_performance(test)

# Affichage des métriques de performance (accuracy, AUC, etc.)
print(perf)
```
