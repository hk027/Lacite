```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

# 📊 Données brutes
data = pd.DataFrame({
    'age': [25, None, 30, 22],
    'salary': [2000, 3000, None, 2500],
    'city': ['Paris', 'Lyon', 'Paris', 'Lyon']
})

print("Données initiales :")
print(data)

# 🔧 1. Gestion des valeurs manquantes
imputer = SimpleImputer(strategy='mean')
data[['age', 'salary']] = imputer.fit_transform(data[['age', 'salary']])

# 🔢 2. Encodage des variables catégorielles
data = pd.get_dummies(data, columns=['city'])

# 📏 3. Normalisation
scaler = StandardScaler()
data[['age', 'salary']] = scaler.fit_transform(data[['age', 'salary']])

print("\nDonnées après préparation :")
print(data)
```

Ce que fait ce code : 

Remplit les valeurs manquantes (moyenne)
Transforme les catégories (Paris/Lyon → 0/1)
Normalise les données
Résultat : données prêtes pour un modèle de
apprentissage automatique
