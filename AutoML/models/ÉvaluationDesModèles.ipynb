```python
# Import des bibliothèques
import h2o
from h2o.automl import H2OAutoML

# 1️⃣ Initialiser H2O
h2o.init()

# 2️⃣ Charger les données (Titanic)
data = h2o.import_file("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv")

# 3️⃣ Définir la variable cible
y = "survived"

# 4️⃣ Définir les variables explicatives
x = data.columns
x.remove(y)

# 5️⃣ Convertir la cible en facteur (classification)
data[y] = data[y].asfactor()

# 6️⃣ Diviser les données (80% train / 20% test)
train, test = data.split_frame(ratios=[0.8], seed=42)

# 7️⃣ Lancer AutoML
aml = H2OAutoML(max_models=10, seed=42)

aml.train(x=x, y=y, training_frame=train)

# 8️⃣ Afficher le leaderboard (classement des modèles)
lb = aml.leaderboard
print(lb)

# 9️⃣ Meilleur modèle
leader = aml.leader

# 🔟 Évaluation sur le test set
perf = leader.model_performance(test)

# Afficher les métriques principales
print("AUC :", perf.auc())
print("Accuracy :", perf.accuracy())
print("F1-score :", perf.F1())
```
