<b>Ce que fait ce code : </b>
 <br>
Démarre H2O et importe les données. <br>
Sélectionne les colonnes prédicteurs et la cible.<br>
Convertit les colonnes catégorielles en facteurs.<br>
Remplit les valeurs manquantes numériques par la médiane.<br>
Sépare le dataset en train/test.<br>
Entraîne un Random Forest pour tester les données préparées.<br>
Évalue les performances sur le test set.<br>

```python
import h2o
from h2o.estimators import H2ORandomForestEstimator

# 1️⃣ Démarrer H2O
h2o.init()

# 2️⃣ Charger le dataset Titanic
data = h2o.import_file("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv")

# 3️⃣ Définir cible et prédicteurs
y = "survived"
X = data.columns
X.remove(y)

# 4️⃣ Convertir la cible en facteur
data[y] = data[y].asfactor()

# 5️⃣ Convertir les colonnes catégorielles en facteurs
categorical_columns = ["sex", "embarked", "class", "who", "deck", "embark_town", "alive"]
for col in categorical_columns:
    if col in X:
        data[col] = data[col].asfactor()

# 6️⃣ Imputer les colonnes numériques sans réassigner
numeric_columns = [col for col in X if data[col].isnumeric()]
for col in numeric_columns:
    data[col].impute(method="median")  # pas de réassignation

# 7️⃣ Séparer train/test
train, test = data.split_frame(ratios=[0.8], seed=1234)

# 8️⃣ Créer et entraîner le Random Forest
model = H2ORandomForestEstimator(ntrees=50, max_depth=20, seed=1234)
model.train(x=X, y=y, training_frame=train)

# 9️⃣ Évaluer les performances
perf = model.model_performance(test_data=test)
print("AUC :", perf.auc())
print("Accuracy :", perf.accuracy()[0][1])
print("F1-score :", perf.F1()[0][1])
print("Matrice de confusion :")
print(perf.confusion_matrix())

# 🔹 Ne pas fermer H2O dans Jupyter pour continuer à travailler
# h2o.shutdown(prompt=False)
```
