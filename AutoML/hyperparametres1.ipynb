```python
# Import des librairies
import pandas as pd
from tpot import TPOTClassifier
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris  # exemple simple si tu n'as pas encore de dataset

# Exemple de dataset (à remplacer par ton DataFrame 'data')
# Ici on crée un DataFrame fictif similaire à ton exemple
# data = pd.read_csv('ton_fichier.csv')  # si tu as un CSV
iris = load_iris()
data = pd.DataFrame(iris.data, columns=iris.feature_names)
data['salary'] = iris.target  # juste pour avoir une colonne cible

# X = variables, y = cible
X = data.drop('salary', axis=1)
y = (data['salary'] > 0).astype(int)  # classification binaire

# Séparer les données en train / test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Créer un modèle TPOT
tpot = TPOTClassifier(
    generations=5,           # Nombre de générations pour l'évolution
    population_size=20,      # Taille de la population
    verbose=2,             # Affiche les détails de l'évolution
    max_time_mins=1          # Temps maximum (en minutes) pour trouver le meilleur pipeline
)

# Entraîner TPOT
tpot.fit(X_train, y_train)

# Évaluer la performance
print("Score :", tpot.score(X_test, y_test))

# Exporter le pipeline final
tpot.export('tpot_pipeline.py')
```
