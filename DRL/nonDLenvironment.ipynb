### Exemple simple en Python (environnement jouet) ###

```python
import random
```

```python
states = ["A", "B", "C"]
actions = ["gauche", "droite"]

state = random.choice(states)

def reward(state, action):
    if state == "C" and action == "droite":
        return 1
    return 0

for step in range(5):
    action = random.choice(actions)
    r = reward(state, action)
    print(f"État: {state}, Action: {action}, Récompense: {r}")
    state = random.choice(states)
```
