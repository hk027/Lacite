```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
```

```python
class Discriminator(nn.Module):  # Définition du discriminateur
    def __init__(self, channels_img=3, features_d=64):  # Constructeur
        super(Discriminator, self).__init__()

        self.net = nn.Sequential(  # Réseau séquentiel

            # --------- Étape 1 : image → features ---------
            nn.Conv2d(channels_img, features_d, 4, 2, 1),  
            # Réduit la taille (64x64 → 32x32)

            nn.LeakyReLU(0.2),  
            # Activation (évite les neurones morts)

            # --------- Étape 2 ---------
            nn.Conv2d(features_d, features_d * 2, 4, 2, 1),  
            # 32x32 → 16x16

            nn.BatchNorm2d(features_d * 2),  
            nn.LeakyReLU(0.2),

            # --------- Étape 3 ---------
            nn.Conv2d(features_d * 2, features_d * 4, 4, 2, 1),  
            # 16x16 → 8x8

            nn.BatchNorm2d(features_d * 4),
            nn.LeakyReLU(0.2),

            # --------- Étape 4 ---------
            nn.Conv2d(features_d * 4, features_d * 8, 4, 2, 1),  
            # 8x8 → 4x4

            nn.BatchNorm2d(features_d * 8),
            nn.LeakyReLU(0.2),

            # --------- Étape finale : décision ---------
            nn.Conv2d(features_d * 8, 1, 4, 1, 0),  
            # 4x4 → 1x1 (score final)

            nn.Sigmoid()  
            # Sortie entre 0 et 1 (probabilité réel/faux)
        )

    def forward(self, x):  
        return self.net(x).view(-1)  
        # Aplati la sortie pour obtenir un vecteur de probabilités
```
