```python
import librosa
#Librosa est une puissante bibliothèque Python pour l'analyse musicale et audio, 
# largement utilisée pour des tâches de recherche d'informations musicales (MIR), 
# telles que l'extraction de caractéristiques audio, la visualisation des formes 
# d'onde et le traitement avancé du signal.
import numpy as np
import matplotlib.pyplot as plt
```

### Charger un fichier audio (mono) ###

```python
# Charger un fichier audio (mono)
y, sr = librosa.load("harvard.wav", sr=None)

print("Fréquence d'échantillonnage :", sr)
print("Nombre total d'échantillons :", len(y))
```

### Extraire une frame audio (25 ms)

```python
frame_duration = 0.025  # 25 ms
N = int(sr * frame_duration)

frame = y[:N]  # première frame
```

### Créer et appliquer la fenêtre de Hamming

```python
# Fenêtre de Hamming
hamming_window = np.hamming(N)

# Application de la fenêtre
windowed_frame = frame * hamming_window
```

### Créer et appliquer la fenêtre de Hamming

```python
# Fenêtre de Hamming
hamming_window = np.hamming(N)

# Application de la fenêtre
windowed_frame = frame * hamming_window
```

### Visualisation : avant / après Hamming

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.plot(frame)
plt.title("Frame audio brute")
plt.xlabel("Échantillons")
plt.ylabel("Amplitude")

plt.subplot(1,2,2)
plt.plot(windowed_frame)
plt.title("Frame avec fenêtre de Hamming")
plt.xlabel("Échantillons")

plt.tight_layout()
plt.show()
```

Ce que tu dois observer<br>

 Discontinuités possibles <br>

🔹 Frame avec Hamming
