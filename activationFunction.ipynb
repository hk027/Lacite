```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
# Define the activation functions 
def linear(x): 
    return x 

#sigmoid 
def sigmoid(x):
    return 1 / (1 + np.exp(-x))


## Completer les ces trois fonctions d'activation (thanh(x),relu(x),softmax(x))et comparer les graphiques 
#thanh(x)
#def tanh(x):
    

#def relu(x): 
    

#def softmax(x): 
    

```

```python
#Generate data 
x = np.linspace(-10, 10, 400) 
```

```python
# Plotting the graphs for each activation function 
fig, axs = plt.subplots(3, 2, figsize=(12, 12)) 

 
# Linear Activation 
axs[0, 0].plot(x, linear(x)) 
axs[0, 0].set_title("Linear Activation: f(x) = x") 
axs[0, 0].set_xlabel("x") 
axs[0, 0].set_ylabel("f(x)") 

#Sigmoid Activation 
axs[0, 1].plot(x, sigmoid(x)) 
axs[0, 1].set_title("Sigmoid Activation: f(x) = 1 / (1 + e^(-x))") 
axs[0, 1].set_xlabel("x") 
axs[0, 1].set_ylabel("f(x)") 

# Tanh Activation 


# ReLU Activation 

# Softmax Activation (for a vector of values) 


#Remove the empty subplot 
axs[2, 1].axis("off") 

# Adjust layout 
plt.tight_layout() 
plt.show() 
```
