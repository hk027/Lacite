```python
import torch.nn as nn
import torch.optim as optim
```

```python
class GRUModel(nn.Module):
    def __init__(self, input_size, hidden_size):
        super().__init__()
        self.gru = nn.GRU(
            input_size=input_size,
            hidden_size=hidden_size,
            batch_first=True
        )
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        # x : (batch_size, sequence_length, input_size)
        out, h_n = self.gru(x)
        # out : (batch_size, sequence_length, hidden_size)

        out = out[:, -1, :]     # (batch_size, hidden_size)
        out = self.fc(out)      # (batch_size, 1)
        return out


model_gru = GRUModel(input_size, hidden_size)
criterion = nn.MSELoss()
optimizer = optim.Adam(model_gru.parameters(), lr=learning_rate)

# Entraînement
for epoch in range(epochs):
    optimizer.zero_grad()
    
    y_pred = model_gru(X_torch)    # (batch_size, 1)
    loss = criterion(y_pred, y_torch)
    
    loss.backward()
    optimizer.step()
    
    if (epoch+1) % 50 == 0:
        print(f"GRU Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")

print("\nExemple de prédiction GRU :")
print("y réel :", y_torch[:3].squeeze())
print("y prédit :", model_gru(X_torch)[:3].detach().squeeze())
```
