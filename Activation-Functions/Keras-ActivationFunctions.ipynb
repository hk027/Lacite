```python
# chapter1/activation_keras.py
import keras
from keras import layers
```

```python
# Define input tensor
x = layers.Input(shape=(10,))
# Sigmoid.
sigmoid = keras.activations.sigmoid(x)

# Tanh.
tanh = keras.activations.tanh(x)

# ReLU.
relu = keras.activations.relu(x)

# LeakyReLU.
lkrelu = keras.layers.LeakyReLU(alpha=0.3)
# Note : ceci est une couche, le paramètre alpha représente
# le gradient lorsque l'entrée est inférieure à 0.

# PReLU.
prelu = keras.layers.PReLU(alpha_initializer='zeros')
# Note : ceci est une couche, le paramètre alpha_initializer
# représente la valeur d'initialisation du gradient lorsque
# l'entrée est inférieure à 0.
 
```
